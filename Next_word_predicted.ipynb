````python
# import libraries
import numpy as np
import pandas as pd
import re
import nltk
from nltk.stem import WordNetLemmatizer
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras import layers
````

# Load a Datasets


````python
with open("/content/1661-0.txt", "r", encoding="utf-8") as file:
    text = file.read()
````

# Text Preprocessing

````python
text = text.lower() # convert to the lower text
matches = re.findall(r'[a-z\s]',text) # select the a -- z and spaces Numbers & Symbols Remove
text = ''.join(matches).strip() # Now Cleaned and text ready!
````

````python
nltk.download('wordnet') # wordnet language database download
lema = WordNetLemmatizer()
````

````python
lem_text = ' '.join([lema.lemmatize(word) for word in text.split()])
````

````python
tok = Tokenizer()
from nltk.tokenize import sent_tokenize


nltk.download('punkt_tab')
text_1 = sent_tokenize(lem_text)
````

````python
tok.fit_on_texts(text_1)

seq = tok.texts_to_sequences(text_1)
````

````python
tok_word_seq = tok.word_index
````

````python
total_words = len(tok.word_index) + 1
tok_word_seq = tok.word_index

print("Total unique words:", total_words)
````

````python
data = [] # empty list
limit = 5000

for s in seq:
    limited_s = s[:limit]

    for i in range(len(limited_s) - 2):
        data.append(limited_s[0:i + 3])
````

````python
# chck the empty list
data
````

````python
# make a x and y empty list
x = []
y = []

for r in data:
    x.append(r[:-1]) # This means: take everything except the last item.
    y.append(r[-1]) # iska mtlb only last item
````

````python
x = pad_sequences(x, maxlen=251, padding='post')
y = np.array(y).reshape(-1, 1)
````

````python
x = np.array(x)
y = np.array(y)

````

````python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.23, random_state=42)
````

````python
model = keras.Sequential([
    layers.Input(shape=(251,1)),
    layers.LSTM(512,return_sequences=True),
    layers.LSTM(256),
    layers.Dense(8646,activation="softmax")
])
````

````python
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
````

````python
model.summary()
````

````python
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=1)
````

````python
def text_predicted(seed_text):
# lower
    seed_text = seed_text.lower()

    # convert word to numerical
    token_list = tok.texts_to_sequences([seed_text])[0]

    token_list = pad_sequences(
        [token_list],
        maxlen=251,
        padding="post"
    )

    predicted_index = np.argmax(
        model.predict(token_list, verbose=0)
    )

    predicted_word = ""

    for word, index in tok.word_index.items():
        if index == predicted_index:
            predicted_word = word
            break

    return predicted_word
````

````python
tex_as_input = input("enter a text: ")

pred = text_predicted(tex_as_input)

print("Next Word Pred: ", {tex_as_input + pred})
````
